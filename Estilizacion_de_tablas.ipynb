## Misión 1: Equipo de Ventas

Necesitamos construir una visualización para indicar los **10 principales clientes por total de ventas**.

Con ello, el equipo de ventas puede pensar en estrategias para mantener a estos clientes comprando en nuestra tienda.

```python
ventas.nlargest(10, columns="ventas")
```

```python
clientes = ventas.groupby("nombre_cliente")["ventas"].sum().nlargest(10).copy()
clientes

#nlargest : ordena de mayor a menor, segun una columna(s), por defecto muestra 5 filas
#copy() : genera una copia del DataFrame original y hacer esa operacion, se recomienda hacer en filtro o slice
```

```python
clientes = clientes.reset_index()
clientes

```

```python
clientes.index
```

```python
# dic = {"nombre_cliente":"Clientes", "ventas":"Ventas"}
# clientes = clientes.rename(columns=dic)
# esto es si quieres cambiar columnas pero solo especificando cuales, y no poniendo todo en secuencia como abajo

clientes.columns = ["Clientes", "Ventas"]  #cambiamos los nombres de las columnas, es por orden (todas columnas)
clientes["Ranking"] = clientes.index + 1   #creamos una columna con los valores del index predetemrinado
clientes = clientes.set_index("Ranking")   #convertir en un indice la columna "Ranking"
clientes
```

Objeto style

```python
estilo = clientes.style   #creamos un objeto tipo style apartir de un DataFrame, para usar sus metodos
estilo                    # y estilizar la tabla que guardamos en una variable estilo
```

```python
estilo.format({"Ventas":"{:,.2f}"}) #aplicamos un modificacion de visualizacion a los valores de Vnetas
estilo                              #aplique "," para separa miles y "." para los decimales con 2 decimales (2f)
                                    #ya lo aplica directamente
```

# Misión 2: Equipo Comercial

Necesitamos construir una visualización que permita asociar el **total de ganancias en ventas y el total de lucros por tipo de producto**.

Ello permitirá que el equipo comercial pueda enfocarse en estrategias para apalancar los ingresos que se reciben con cada tipo de producto vendido en la tienda.

# Misión 3: Equipo de Logística

Necesitamos construir una visualización que nos permita entender la **distribución de pedidos por región de Brasil**.

Así, el equipo de logística va a saber dónde hay más oportunidad de enfocar los esfuerzos de material y mano de obra.

```python
ventas.head()
```

```python
region = pd.DataFrame(ventas["region"].value_counts())  #convertimos en un DataFrame esa serie
                                                        #donde mmuestra la cantidad de registros agrupados
region.columns = ["Nº Pedidos"]
region.index.name = "Region"
region
```

Vamos a agrega runa columna de porcentaje de cada valor sobre el total, para ello utilizaremos numpy

```python
import numpy as np
#para realizar operaciones en un array con varias datos, mas optimo
```

```python
porcentaje = region["Nº Pedidos"].to_numpy()   #convertimos en un numpy array esa serie
porcentaje = 100*porcentaje/porcentaje.sum()      #aplicamos formula porcentaje a todos valores del array
porcentaje
```

```python
region["Porcentaje de Pedidos"] = porcentaje                #creamos una columna con los valores del array que operamos
region
```

Vamos a estilizarlo para que se vea mejor

```python
s_region = region.style
s_region
```

```python
encabezado = {
    "selector": "th",
    "props": "font-family:Arial; font-weight:bold; text-align:right; background-color:white; color:black"
}
# th : afecta a las encabezados y al indice como a sus valores del indice

celdas = {
    "selector": "td",
    "props": "text-align:right; background-color:white; color:black"
}
#td : afecta a los valores de las columnas, contenido

s_region.set_table_styles([encabezado, celdas])
```

```python
s_region.format({"Porcentaje de Pedidos": "{:.2f} %"})\
                  .bar(subset="Porcentaje de Pedidos", vmin=0, vmax=100, color="#9cd3bb")
```

Otras funciones built-in
- highlight_null : resaltar valores nulos
- highlight_between : resaltar valores en un rango
- highlight_quantile : resaltar rango de cuantil especifico
- text_gradient : es como background_gradiente pero para el texto

Ejemplo con los dos primeros

```python
import pandas as pd
import numpy as np

datos = pd.DataFrame({'ID': [1204, 3018, 9963, 1005, 5627],
                   'Edad': [np.nan, 32, 19, np.nan, 38],
                   'Salario': [3500.00, 4200.50, 2800.00, 5700.25, 4900.75]})
```

```python
s_datos = datos.style.format({"Edad": "{:.0f}", "Salario": "PEN {:,.2f}"}).highlight_null(color="#F79747")
s_datos
```

```python
s_datos.highlight_between(left=4000, right=5000, axis=1, color='#94B2D7')
#se aplico limites directos (numeros), la cual afecta a todos los valores de la columnas (axis 1)
```

```python
s_datos.highlight_between(subset="Salario",left=4000, right=5000, axis=1, color='#94B2D7')
#aplica solo a la columna (axis 1) Salario
```

```python
limite_inferior = pd.Series([4000.0], index=['Salario'])    # 1 forma de crear una serie, es para indicar en que columna se realizara
limite_inferior = pd.Series({"Salario" : 4000, "Edad":0})   # 1 forma tambien de crear una serie,

limite_superior = pd.Series({"Salario": 5000, "Edad":20})   #especificar el limite para cada columna
                                                            #se puede poner una lista simple con numeros en secuencia, pero es mas elegante asi

```

```python
s_datos.highlight_between(left=limite_inferior, right=limite_superior, axis=1, color='#94B2D7')
#se aplica una serie como limites, que indican en que columna se haran, pero se aplica un solo color
```

Si quieres aplicar varios filtros  pero con diferentes colores, aplicamos una secuencia de funciones

```python
s_datos.highlight_between(subset="Edad",left=0, right=20, color="lightblue").highlight_between(subset="Salario", left=4000, right=5000, color="lightgreen")
```

Si quieres aplicar varios filtros con highlight_between y con diferentes colores, aplicamos un diccionario con los parametros, para que sea mas ordenado

```python
reglas = {
    'Edad': {'left': 0, 'right': 20, 'color': 'lightblue'},
    'Salario': {'left': 4000, 'right': 5000, 'color': 'lightgreen'}
}

# Aplicamos todo dinámicamente
for col, params in reglas.items():
    s_datos.highlight_between(subset=col, **params)

s_datos
```

```python
df_descuentos = tienda_libros["tipo_descuento"].value_counts().to_frame()
df_descuentos
#convertimos a DataFrame
```

```python
df_descuentos.index.name = "Tipo de descuento"
df_descuentos.columns = ["Cantidad de pedidos"]
df_descuentos
#Cambiamos los nombes del index y columna
```

```python
import numpy as np
```

```python
porcentaje = df_descuentos["Cantidad de pedidos"].to_numpy()
porcentaje = 100*porcentaje/porcentaje.sum()
porcentaje
#creamos un numpy array donde estaran los porcentajes para valores de nueva columna
```

```python
df_descuentos["Porcentaje de pedidos"] = porcentaje
df_descuentos
#creamos una nueva columna con los valores de porcetaje (numpy array)
```

```python
s_descuentos = df_descuentos.style
s_descuentos.format({"Porcentaje de pedidos": "{:.2f} %"})\
                  .bar(subset="Porcentaje de pedidos",vmin=0, vmax=100, color="#093364", axis=1,
                       width=60, height=50)
#creamos una mascara de estilos a nuestro DataFrame
#aplicamos un formato a los valores de porcentaje
#agregas un grafico de barras a la columna porcentajes, aplicamos color, limites referencia
#ancho y altura
```
